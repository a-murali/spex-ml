```python
flux_vals = pd.DataFrame(singles['FLUX'].to_list(), columns=[f'FLUX_{i}' for i in range(409)])
flux_vals
```

```python
# PER-ROW NORMALIZATION TO 1.25 micrometer FLUX VALUE (OR SHOULD WE DO NORMALIZATION TO AVERAGE OF 1.25 micrometer FLUX VALUE?)
flux_vals_normalized = flux_vals.apply(lambda row: row / row.iloc[88], axis=1)
```

```python
# PER-ROW 0-1 NORMALIZATION (MAX-MIN NORMALIZATION USING PER-ROW MAX FLUX AND MIN FLUX)
flux_vals_row_max_min_norm = flux_vals.apply(lambda row: (row - row.min()) / (row.max() - row.min()), axis=1)
flux_vals_row_max_min_norm.describe()
```

```python
#OVERALL 0-1 NORMALIZATION (USING MAX OVERALL FLUX AND MIN OVERALL FLUX)
max_flux = flux_vals.max().max()
min_flux = flux_vals.min().min()
flux_vals_overall_max_min_norm = flux_vals.apply(lambda row: (row - min_flux) / (max_flux - min_flux), axis=1)
flux_vals_overall_max_min_norm.describe()
```

## PCA

```python
#NORMALIZE DATA FOR PCA
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
flux_vals_rescaled = scaler.fit_transform(flux_vals_normalized)
pd.DataFrame(flux_vals_rescaled).describe()
```

```python
#PCA
from sklearn.decomposition import PCA

pca = PCA(n_components = 0.97)
pca.fit(flux_vals_rescaled)
reduced_flux_vals = pca.transform(flux_vals_rescaled)
reduced_flux_vals = pd.DataFrame(reduced_flux_vals)

reduced_flux_vals.describe()
```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
```

```python
new_spt = singles["SPT"].str[0]
singles["SPT"] = new_spt
```

```python
singles_drop = singles["SPT"]
singles_w_flux_vals = pd.concat([singles_drop, reduced_flux_vals], axis = 1)
singles_w_flux_vals.head()
```

```python
X = singles_w_flux_vals.drop("SPT", axis = 1)
y = singles_w_flux_vals["SPT"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 42)

rf_classifier = RandomForestClassifier()
rf_classifier.fit(X_train, y_train)
preds = rf_classifier.predict(X_test)
preds[:5]
```

```python
ACC = np.mean(preds == y_test)

print(f"RF ACCURACY: {ACC}")
```

## Hyperparameter Tuning (Grid Search)

```python
from sklearn.model_selection import GridSearchCV
```

```python
y_train.value_counts()
```

```python
rf_parameters = {'n_estimators': [50, 100, 200, 500],
                 'min_samples_split': [2, 5, 10],
                 'min_samples_leaf': [1, 2, 5, 10],
                 'max_depth' : [5, 7, 9, 10]}

CV_rfc = GridSearchCV(estimator = rf_classifier, param_grid = rf_parameters, scoring = 'accuracy', cv = 5)
CV_rfc.fit(X_train, y_train)
```

```python
CV_rfc.best_params_
```

```python
rf_classifier = RandomForestClassifier(max_depth = 10, min_samples_leaf = 1, min_samples_split = 2, n_estimators = 100)
rf_classifier.fit(X_train, y_train)
preds = rf_classifier.predict(X_test)
preds[:5]
```

```python
ACC = np.mean(preds == y_test)

print(f"RF ACCURACY: {ACC}")
```
